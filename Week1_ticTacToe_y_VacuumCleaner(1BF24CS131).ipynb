```python
class VacuumStateSpace:
    def __init__(self, initial_location='A', status_A='Dirty', status_B='Dirty'):
        self.state = (initial_location, status_A, status_B)
        self.path_cost = 0
        self.steps = []

    def get_state(self):
        return self.state

    def goal_test(self):
        agent_loc, status_A, status_B = self.state
        return status_A == 'Clean' and status_B == 'Clean'

    def transition(self, action):
        agent_loc, status_A, status_B = self.state
        next_loc, next_A, next_B = agent_loc, status_A, status_B

        if action == "Left":
            if agent_loc == 'B':
                next_loc = 'A'

        elif action == "Right":
            if agent_loc == 'A':
                next_loc = 'B'

        elif action == "Pick Dust":
            if agent_loc == 'A' and status_A == 'Dirty':
                next_A = 'Clean'
            elif agent_loc == 'B' and status_B == 'Dirty':
                next_B = 'Clean'

        self.state = (next_loc, next_A, next_B)
        self.path_cost += 1
        self.steps.append(action)

vacuum = VacuumStateSpace(initial_location='A', status_A='Dirty', status_B='Dirty')

print(f"Initial State: {vacuum.get_state()}")

plan = ["Pick Dust", "Right", "Pick Dust"]

for action in plan:
    vacuum.transition(action)
    print(f"Action: {action:<11}\n-------------------------\n\nNew State: {vacuum.get_state()} | Goal Reached: {vacuum.goal_test()}")

print(f"\nTotal Path Cost: {vacuum.path_cost}")
print(f"Execution Path: {' -> '.join(vacuum.steps)}")
```

```python
import random

def print_board(board):
    print("\n")
    for i, row in enumerate(board):
        # Format None values to display as empty spaces
        row_display = [cell if cell is not None else " " for cell in row]
        print(f" {row_display[0]} | {row_display[1]} | {row_display[2]} ")
        if i < 2:
            print("---+---+---")
    print("\n")

def check_winner(board):
    # Check Rows
    for row in board:
        if row[0] == row[1] == row[2] and row[0] is not None:
            return row[0]
    # Check Columns
    for col in range(3):
        if board[0][col] == board[1][col] == board[2][col] and board[0][col] is not None:
            return board[0][col]
    # Check Diagonals
    if board[0][0] == board[1][1] == board[2][2] and board[0][0] is not None:
        return board[0][0]
    if board[0][2] == board[1][1] == board[2][0] and board[0][2] is not None:
        return board[0][2]
    return None

def get_empty_cells(board):
    """Returns a list of coordinate tuples (r, c) that are empty."""
    empty = []
    for r in range(3):
        for c in range(3):
            if board[r][c] is None:
                empty.append((r, c))
    return empty





# Initialize a 3x3 matrix with None
board = [[None for _ in range(3)] for _ in range(3)]
user = 'X'
computer = 'O'
current_turn = 'X'  # User goes first

print("Welcome to Tic-Tac-Toe! You are 'X' and the Computer is 'O'.")
print_board(board)

# --- Main Game Loop ---
while True:
    empty_cells = get_empty_cells(board)

    # Check for a Tie (no empty spaces left)
    if not empty_cells:
        print("It's a tie game!")
        break

    if current_turn == user:
        # --- User's Turn ---
        print("Your turn (X).")
        try:
            row = int(input("Enter row (0, 1, or 2): "))
            col = int(input("Enter column (0, 1, or 2): "))

            if (row, col) in empty_cells:
                board[row][col] = user
                print_board(board)

                # Check if user won
                if check_winner(board) == user:
                    print("Congratulations! You beat the computer!")
                    break

                current_turn = computer  # Switch to computer
            else:
                print("Invalid move! That spot is taken or out of bounds. Try again.")
        except ValueError:
            print("Please enter valid numbers (0, 1, or 2).")

    else:
        # --- Computer's Turn ---
        print("Computer's turn (O)...")
        # Simply pick a random available spot
        row, col = random.choice(empty_cells)
        board[row][col] = computer
        print_board(board)

        # Check if computer won
        if check_winner(board) == computer:
            print("The computer wins! Better luck next time.")
            break

        current_turn = user  # Switch to user
```
